```python
import matplotlib.pyplot as plt
import torch
def plot_channel_heatmaps(tensor_or_array, cmap='viridis'):

    if isinstance(tensor_or_array, torch.Tensor):
        data = tensor_or_array.detach().cpu().numpy()
    else:
        data = np.copy(tensor_or_array)
        
    if data.ndim == 4 and data.shape[0] == 1:
        data = data[0]
        
    num_channels = data.shape[0]
    
    fig, axes = plt.subplots(1, num_channels, figsize=(5 * num_channels, 5))
    
    if num_channels == 1:
        axes = [axes]
        
    for i in range(num_channels):
        im = axes[i].imshow(data[i], cmap=cmap)
        axes[i].set_title(f'Channel {i}')
        axes[i].axis('off')
        
        fig.colorbar(im, ax=axes[i], fraction=0.046, pad=0.04)
        
    plt.tight_layout()
    plt.show()
```

```python
plot_channel_heatmaps(summed_mask)
```
